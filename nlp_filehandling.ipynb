```python
!pip install PyPDF2
from PyPDF2 import PdfReader
from PyPDF2 import PdfWriter
from PyPDF2 import PdfMerger
```

```python
# Read a PDF file

reader = PdfReader('NLP_Learning_Notes.pdf')
print(reader)
```

```python
# Count total number of pages

print(f'Total number of pages are {len(reader.pages)}')
```

```python
# Read first page

page = reader.pages[0]
print(page.extract_text())
```

```python
# Read complete PDF

for page in reader.pages:
  print(page.extract_text())
```

```python
# copy complete PDF

writer = PdfWriter()
for page in reader.pages:
  writer.add_page(page)

with open('NLP_Learning_Notes_copy.pdf', 'wb') as file:
  writer.write(file)
```

```python
# Split PDF

reader = PdfReader('NLP_Learning_Notes.pdf')
writer = PdfWriter()
writer.add_page(reader.pages[0])
with open('NLP_Learning_Notes_page1.pdf', 'wb') as file:
  writer.write(file)
```
