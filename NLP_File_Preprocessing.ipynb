```python
import re
import string
from bs4 import BeautifulSoup
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
sample_text = """
Hey there!!! 😊😊

I'm Ayesha, and I LOVE Machine Learning & NLP!!! 🚀🚀

Visit my website: https://www.mywebsite.com
or http://example.org

Here's another HTML page:
<a href="https://openai.com">OpenAI</a>

<div class="content">
    <h2>Today's News</h2>
    <p>Artificial Intelligence is changing the world in 2026!!</p>
</div>

Email me at: ayesha123@gmail.com

Price of laptop is Rs. 78,999.50 😎
Accuracy = 98.75%
Phone: +91-9876543210

Python, PYTHON, python...
Machine-learning is awesome!!!

I have    multiple      spaces.

Special symbols:
@#$%^&*()_+=[]{}|\\:;"'<>,.?/~`

Happy coding!!! ❤️❤️🔥🔥
"""

with open('data.txt', 'w') as file:
  file.write(sample_text)

print("Data saved successfully.")
```

```python
# read file
with open('data.txt', 'r') as file:
  text = file.read()

print(text)
```

```python
# Remove HTML tags
text = BeautifulSoup(text, 'html.parser').get_text()
print(text)
```

```python
# Remove URL
text = re.sub(r'http\S+|www\S+', '', text)
print(text)
```

```python
# Remove email address
text = re.sub(r'\S+@\S+', '', text)
print(text)
```

```python
# Remove Phone Number
text = re.sub(r'\+?\d\[\d -]{8,}\d', '', text)
print(text)
```

```python
# Remove percentages
text = re.sub(r'\b\d+(\.\d+)?%', '', text)
```

```python
# Remove emojis and other non-ascii characters
text = text.encode('ascii', 'ignore').decode('ascii')
print(text)
```

```python
# convert to lower case
text = text.lower()
print(text)
```

```python
# Remove Numbers
text = re.sub(r"\b\d+([.,]\d+)*%?\b", "", text)
print(text)
```

```python
# Remove _
text = re.sub(r"_", " ", text)
print(text)
```

```python
# Remove punctuation
text = re.sub(r'[^\w\s]', '', text)
print(text)
```

```python
# Remove extra spaces
text = re.sub(r'\s+', ' ', text).strip()
print(text)
```

```python
nltk.download('punkt_tab', quiet=True)
# Tokenization
tokens = word_tokenize(text)
print(tokens)
print("\nTotal Tokens:", len(tokens))
```

```python
# Stopword removal
stop_words = set(stopwords.words("english"))

filtered_tokens = [
    word for word in tokens
    if word not in stop_words
]

print(filtered_tokens)
print("\nTotal Tokens:", len(filtered_tokens))
```

```python
# Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized = [
    lemmatizer.lemmatize(word)
    for word in filtered_tokens
]

print(lemmatized)
```

```python
# Save clean Data
with open('clean_data.txt', 'w') as file:
  file.write(' '.join(tokens))

print("Clean data saved successfully.")
```
