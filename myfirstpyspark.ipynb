```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('My First PySpark').getOrCreate()
## SparkSession - create spark application.
## builder - build spark
## appName - gives name to the app
```

```python
## Create my first data
data = [('John', 25), ('Julia', 27), ('Max', 29), ('Lucas', 32), ('David', 35)]
```

```python
df = spark.createDataFrame(data, ["Name", "Age"])
df
```

```python
df.show(2) # Shows only 2 cells
```

```python
df.printSchema()
```

```python
employee_data = [(101, 'Max', 'AI', 100000),
                 (102, 'Lucas', 'ML', 90000),
                 (103, 'Julia', 'DS', 50000),
                 (104, 'Yana', 'AI', 100000),
                 (105, 'Pablo', 'CV', 80000)]
emp_df = spark.createDataFrame(employee_data, ['Employee ID', 'Employee Name', 'Department', 'Salary'])
emp_df.show()
```

```python
emp_df.show(1)
```

```python
## Check spark version
print(spark.version)
```

```python
emp_df.filter(emp_df.Department == 'AI').show()
```

```python
emp_df.filter(emp_df.Salary < 90000).show()
```

```python
emp_df.filter(emp_df.Department == 'Sales').show()
```

```python
emp_df.filter((emp_df.Salary == 100000) & (emp_df['Employee Name'] == 'Max')).show()
```

```python
from pyspark.sql.functions import col
```

```python
emp_df = emp_df.withColumn("Bonus", col('Salary') + 10000)
```

```python
emp_df.show()
```

```python
emp_df = emp_df.withColumn("Replica_ID", col('Employee ID') + 100)
```

```python
emp_df.show()
```

```python
emp_df = emp_df.drop('Replica_ID')
```

```python
emp_df.show()
```

```python
emp_df.describe().show()
```

```python
## Read Diabetes dataset from CSV.

spark = SparkSession.builder.appName('Diabetes dataset').getOrCreate()
```

```python
df = spark.read.csv('diabetes.csv', header=True, inferSchema=True)
```

```python
df.show()
```

```python
df.printSchema()
```

```python
df.select('Glucose').show()
```

```python
df.filter(df.Age < 50).show()
```

```python
df.filter((df.Age < 30) & (df.Outcome == 1)).show()
```

```python
df.drop('SkinThickness').show()
```

```python
rows = df.take(10)
for row in rows[4:10]:
  print(row)
```

```python
from pyspark.sql.functions import max
```

```python
df.select(max('age')).show()
```

```python
from pyspark.sql.functions import min
df.select(min('age')).show()
```

```python
from pyspark.sql.functions import avg
df.select(avg('Glucose')).show()
```
