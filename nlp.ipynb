```python
from bs4 import BeautifulSoup   # used to remove html and xml tags
from nltk.tokenize import word_tokenize   # makes tokens on words
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer, PorterStemmer
from nltk import pos_tag
import regex as re
!pip install emoji
import emoji
!pip install contractions
import contractions
```

```python
# load Spacy English Model

nlp = spacy.load('en_core_web_sm')
```

```python
# Input Text

text = """
<p>Helloo! 😊 I'm John, and I visited https://openai.com on 12/08/2026.
I can't believe AI is growing so fast!!!</p>
"""

print("="*60)
print("Original Text:\n")
print("="*60)
print(text)
```

```python
# Step 1: Lowercase Conversion

text = text.lower()
print("="*60)
print("Lowercase Conversion:\n")
print("="*60)
print(text)
```

```python
# Step 2: Remove HTML Tags

text = BeautifulSoup(text, 'html.parser').get_text()
print("="*60)
print("Remove HTML Tags:\n")
print("="*60)
print(text)
```

```python
# Step 3: Remove URL

text = re.sub(r'http\S+', '', text)
print("="*60)
print("Remove URL:\n")
print("="*60)
print(text)
```

```python
# Step 4: Remove Emojis

text = emoji.replace_emoji(text, replace='')
print("="*60)
print("Remove Emojis:\n")
print("="*60)
print(text)
```

```python
# Step 5: Expand Contractions

text = contractions.fix(text)
print("="*60)
print("Expand Contractions:\n")
print("="*60)
print(text)
```

```python
# Step 6: Remove Special Characters

text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
print("="*60)
print("Remove Special Characters:\n")
print("="*60)
print(text)
```

```python
# Step 7: Remove numbers

text = re.sub(r'\d+', '', text)
print("="*60)
print("Remove numbers:\n")
print("="*60)
print(text)
```

```python
# Step 8: Remove Extra Space

text = re.sub(r'\s+', ' ', text).strip()
print("="*60)
print("Remove Extra Space:\n")
print("="*60)
print(text)
```

```python
# Step 9: Tokenization
nltk.download('punkt_tab')
tokens = word_tokenize(text)
print("="*60)
print("Tokenization:\n")
print("="*60)
print(tokens)
```

```python
# Step 10: Stopword Removal
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
filtered_tokens = [token for token in tokens if token not in stop_words]
print("="*60)
print("Stopword Removal:\n")
print("="*60)
print(filtered_tokens)
```

```python
# Step 11: Lemmatization
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]
print("="*60)
print("Lemmatization:\n")
print("="*60)
print(lemmatized_tokens)
```

```python
# Step 12: POS Tagging
nltk.download('averaged_perceptron_tagger_eng')
pos_tags = pos_tag(lemmatized_tokens)
print("="*60)
print("POS Tagging:\n")
print("="*60)
print(pos_tags)
```

```python
# Step 13: Named Entity

doc = nlp(" ".join(lemmatized_tokens))

print("="*60)
print("Named Entity:\n")
print("="*60)

for ent in doc.ents:
    print(ent.text, "------->", ent.label_)
```

```python
# Step 14: Bag Of Words

from sklearn.feature_extraction.text import CountVectorizer

clean_text = " ".join(lemmatized_tokens)
corpus = [clean_text]

bow = CountVectorizer()
bow_matrix = bow.fit_transform(corpus)

print("Bag Of Words Matrix:\n")
print(bow_matrix.toarray())

print("\nVocabulary:\n")
print(bow.vocabulary_)
```

```python
# Step 15: TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(corpus)

print("TF-IDF Matrix:\n")
print(tfidf.get_feature_names_out())
print(tfidf_matrix.toarray())
```

```python
# ==========================================================
# STEP 17 : WORD2VEC
# ==========================================================

from gensim.models import Word2Vec

sentences = [lemmatized_tokens]

model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4
)

print("\nVector of 'ai'")

print(model.wv["ai"])
```
