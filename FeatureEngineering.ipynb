```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv('loan_approved.csv')
```

```python
category_columns = ['Married', 'Self_Employed', 'Dependents', 'Gender']

for col in category_columns:
    df[col].fillna(df[col].mode()[0], inplace=True)

df.isnull().sum()
```

```python
## Checking duplicate data

df.duplicated().sum()
```

```python
## Change Data Types just for reference

# df['Age'] = df['Age'].astype('int64')
# df['Age'] = df['Age'].astype('float')
# df['Loan_ID'] = df['Loan_ID'].astype('str')
```

```python
## Handling incorrect data

df['Gender'] = df['Gender'].str.lower()
```

```python
##Loan status vs Gender

sns.countplot(x="Gender", hue="Loan_Status (Approved)", data=df)
plt.show()
```

```python
##Loan status vs Married

sns.countplot(x="Married", hue="Loan_Status (Approved)", data=df)
plt.show()
```

```python
##Loan status vs Property Area

sns.countplot(x="Property_Area", hue="Loan_Status (Approved)", data=df)
plt.show()
```

```python
## Income vs Loan status

plt.figure(figsize=(8,5))
sns.boxplot(x="Loan_Status (Approved)", y="ApplicantIncome", data=df)
plt.title("Applicant Income vs Loan Status")
plt.show()
```

```python
## Violin Plot

plt.figure(figsize=(8,5))
sns.violinplot(x="Loan_Status (Approved)",
               y="ApplicantIncome",
               data=df)

plt.show()
```

```python
## Histogram with Loan Status

plt.figure(figsize=(8,5))
sns.histplot(data=df,
             x="ApplicantIncome",
             hue="Loan_Status (Approved)",
             kde=True)

plt.show()
```

```python
## Label Encoding of Loan_Status (Approved) column

df['Loan_Status (Approved)'] = df['Loan_Status (Approved)'].map({'Y': 1, 'N':0})
```
