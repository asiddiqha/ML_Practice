### Set - No order, no indexing, no duplicates allowed
- Mathematical Operations
- Mutable

```python
s={1,2,3,4,6,6,6,6}
```

```python
s
```

```python
type(s)
```

```python
s
```

```python
#union, intersection and difference
```

```python
s1={2,5,8,9,1,0}
s2={2,4,7,9,1}
```

```python
#union - all
s1.union(s2)
```

```python
s1 | s2
```

```python
#intersection - common
s1.intersection(s2)
```

```python
s1 & s2
```

```python
#difference - all elements of s2 (whatever is there in s1) got removed
s1.difference(s2)
```

```python
s1 - s2
```

```python
s2 - s1
```
